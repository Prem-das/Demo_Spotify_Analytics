```python
import pandas as pd 


df = pd.read_csv('data\spotify_history.csv')

# Convert timestamp column to proper datetime object
df['ts'] = pd.to_datetime(df['ts'])

# 2. Inspect shape and columns
print(f"Total Rows (Streams): {len(df):,}")
print(f"Total Columns: {len(df.columns)}")
print(f"Columns: {list(df.columns)}\n")

# 3. Data Types and Missing Values
print(df.info())
print("\nMissing Values Count:")
print(df.isnull().sum())

# 4. Sample Check 
print(df.head())

# 5. High-Level Summary Statistics
total_hours = df['ms_played'].sum() / 3600000.0
min_date = df['ts'].min().strftime('%Y-%m-%d')
max_date = df['ts'].max().strftime('%Y-%m-%d')
unique_artists = df['artist_name'].nunique()


print(f"Date Range: {min_date} to {max_date}")
print(f"Total Listening Duration: {total_hours:,.2f} Hours")
print(f"Unique Artists: {unique_artists:,}")
```

```python
# Cleaning Nulls
df[['reason_start', 'reason_end']] = df[['reason_start', 'reason_end']].fillna('unknown')
```

```python
# Checking for Nulls
print(df.isnull().sum())
```

```python
# --- 2. Duration Conversions ---
df['minutes_played'] = df['ms_played'] / 60000.0
df['hours_played'] = df['ms_played'] / 3600000.0
```

```python
print(f"Columns: {list(df.columns)}\n")
```

```python
# Exporting To SSMS 
import pandas as pd
from sqlalchemy import create_engine

server = r"SERVERNAME\SQLEXPRESS"
database = "Spotify"

connection_string = (
    f"mssql+pyodbc://@{server}/{database}"
    "?driver=ODBC+Driver+18+for+SQL+Server"
    "&trusted_connection=yes"
    "&TrustServerCertificate=yes"
)

engine = create_engine(connection_string)
```

```python
with engine.connect() as conn:
    print("Connected successfully!")
```

```python
df.to_sql(
    "spotify_cleaned_dataset",
    con=engine,
    schema="dbo",
    if_exists="replace",
    index=False)
```
